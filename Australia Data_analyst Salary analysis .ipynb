```python
import matplotlib.pyplot as plt

df.groupby('experience_level')['salary_in_usd'].mean().plot(kind='bar')

plt.title('Average Salary by Experience Level')
plt.xlabel('Experience Level')
plt.ylabel('Average Salary (USD)')
plt.show()                                                                                                                                                                                                                                                                                                                                                                                   

```

```python
plt.hist(df['salary_in_usd'],bins=20)
plt.title('Salary Distribution')
plt.xlabel('Salary(USD)')
plt.ylabel('Number of Employees')
plt.show()
```

```python
df.boxplot(column = 'salary_in_usd',
by='experience_level')

plt.title('Salary Distribution by Experience Level')
plt.suptitle('')
plt.xlabel('Experience Level')
plt.ylabel('salary(USD)')
plt.show()
```

plt.figure(figsize = (10,5))


```python
plt.figure(figsize=(10, 5))

df['employment_type'].value_counts().plot(kind='bar', color='skyblue', edgecolor='black')

plt.title('Employee Distribution by Employment Type')
plt.xlabel('Employment Type')
plt.ylabel('Number of Employees')
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 5))

year_salary = df.groupby('work_year')['salary_in_usd'].mean()

plt.plot(year_salary.index, year_salary.values,
         marker='o', linestyle='-', label='Average Salary')

plt.title('Average Salary Trend by Year')
plt.xlabel('Work Year')
plt.ylabel('Average Salary (USD)')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

```python
top_jobs = df.groupby('job_title')['salary_in_usd'].mean().nlargest(10)

plt.figure(figsize=(12, 7))

bars = plt.barh(top_jobs.index, top_jobs.values,
                color='skyblue', edgecolor='black')

plt.title('Top 10 Job Titles by Average Salary')
plt.xlabel('Average Salary (USD)')
plt.ylabel('Job Title')

plt.bar_label(bars, fmt='$%.0f', padding=5)
plt.grid(axis='x', linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```
